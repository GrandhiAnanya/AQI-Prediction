```python
import pandas as pd

input_file = r"Datasets\Combined_Stations_Fixed.csv"

df = pd.read_csv(input_file)
```

```python
# Rows missing at least one value anywhere
df_dropna = df.dropna()

# Rows missing either future target
df_targets = df.dropna(
    subset=["AQI_future_1", "AQI_future_24"]
)

print("Original rows:", len(df))

print("After full dropna:", len(df_dropna))
print("Rows removed by full dropna:", len(df) - len(df_dropna))

print("\nAfter dropping missing targets:", len(df_targets))
print("Rows removed by target filtering:", len(df) - len(df_targets))

print("\nAdditional rows removed due to missing features:",
      len(df_targets) - len(df_dropna))
```

```python
cols_to_remove = [
    col for col in df.columns
    if "_roll_" in col
    and (col.endswith("_1") or "_1_" in col)
]

df_new = df.drop(columns=cols_to_remove)
```

```python
missing_report = pd.DataFrame({
    "Column": df_new.columns,
    "Missing_Count": df_new.isnull().sum().values,
    "Missing_Percentage": (df_new.isnull().mean().values * 100).round(2)
})

# Sort by missing count (highest first)
missing_report = missing_report.sort_values(
    by="Missing_Count",
    ascending=False
).reset_index(drop=True)

print(f"Dataset shape: {df_new.shape[0]:,} rows × {df_new.shape[1]:,} columns")
print("\nMissing values by column:")
print(missing_report.to_string(index=False))

# Save report
missing_report.to_csv("missing_values_report.csv", index=False)

print("\nReport saved as missing_values_report.csv")
```
